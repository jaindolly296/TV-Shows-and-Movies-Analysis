# Step 10: Remove Duplicates

```python
# Remove duplicates
titles.drop_duplicates(inplace=True)

credits.drop_duplicates(inplace=True)

# Verify
print("Titles Shape:", titles.shape)
print("Credits Shape:", credits.shape)
```

Meaning

Deletes duplicate records permanently.

# Step 11: Check Missing Values

```python
# Count missing values
titles.isnull().sum().sort_values(ascending=False)

```

```python
# Count missing values
credits.isnull().sum().sort_values(ascending=False)
```

# Step 12: Fill Description Null Values

```python
# Replace missing descriptions
titles['description'] = titles['description'].fillna('Description Not Available')
```

# Step 13: Fill Age Certification

```python
titles['age_certification'] = titles['age_certification'].fillna('Unknown')
```

# Step 14: Fill IMDb Score

```python
# Use median because ratings may contain outliers
titles['imdb_score'] = titles['imdb_score'].fillna(
    titles['imdb_score'].median()
)
```

# Step 15: Fill IMDb Votes

```python
titles['imdb_votes'] = titles['imdb_votes'].fillna(
    titles['imdb_votes'].median()
)
```

# Step 16: Fill TMDB Popularity

```python
titles['tmdb_popularity'] = titles['tmdb_popularity'].fillna(
    titles['tmdb_popularity'].median()
)
```

# Step 17: Fill TMDB Score

```python
titles['tmdb_score'] = titles['tmdb_score'].fillna(
    titles['tmdb_score'].median()
)
```

# Step 18: Handle Seasons Column

Most nulls occur because movies don't have seasons.

```python
titles['seasons'] = titles['seasons'].fillna(0)
```

# Step 19: Handle IMDb ID

```python
titles['imdb_id'] = titles['imdb_id'].fillna('Not Available')
```

# Step 20: Credits Dataset Cleaning

The only column with many nulls:

```python
credits.isnull().sum()
```

# Step 21: Fill Character Column

```python
credits['character'] = credits['character'].fillna('Unknown Character')
```

# Step 28: Merge Both Datasets

```python
prime_df = pd.merge(
    titles,
    credits,
    on='id',
    how='left'
)
```

Meaning

Join title information with actor/director information.

# Step 29: Final Check

```python
prime_df.info()
```

```python
prime_df.isnull().sum()
```

```python
# Percentage of missing values
missing_percent = (prime_df.isnull().sum() / len(prime_df)) * 100

print(missing_percent.sort_values(ascending=False))
```

```python
# Character name missing
prime_df['character'] = prime_df['character'].fillna('Unknown Character')

# Person name missing
prime_df['name'] = prime_df['name'].fillna('Unknown')

# Person ID missing
prime_df['person_id'] = prime_df['person_id'].fillna(-1)

# Role missing
prime_df['role'] = prime_df['role'].fillna('Unknown Role')
```

# Step 1: Content Type Distribution
Question:

How many Movies and TV Shows are available?

```python
import os
import matplotlib.pyplot as plt

# Create folder (runs only once)
os.makedirs("Amazon_Prime_Visualizations", exist_ok=True)

prime_df['type'].value_counts().plot(kind='bar')

plt.title('Movies vs Shows')
plt.xlabel('Content Type')
plt.ylabel('Count')

# Save image
plt.savefig(
    "Amazon_Prime_Visualizations/Movies_vs_Shows.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

**Insight**

Which content dominates Amazon Prime?

Movies usually outnumber shows.

# Step 2: Content Release Trend
Question:

How has content production changed over time?

```python
import os
import matplotlib.pyplot as plt

# Create folder if it doesn't exist
os.makedirs("Amazon_Prime_Visualizations", exist_ok=True)

# Create the data
release_trend = prime_df.groupby('release_year')['id'].count()

# Plot
release_trend.plot(figsize=(12,5), color='steelblue')

# Titles and labels
plt.title('Content Released Over Time')
plt.xlabel('Year')
plt.ylabel('Number of Titles')

# Save the image
plt.savefig(
    "Amazon_Prime_Visualizations/Content_Released_Over_Time.png",
    dpi=300,
    bbox_inches='tight'
)

# Show the plot
plt.show()
```

**Insight**

Growth trend of content library.

Peak production years.

# Step 3: IMDb Score Distribution
Question:

How are ratings distributed?

```python
import os
import matplotlib.pyplot as plt

# Create folder if it doesn't exist
os.makedirs("Amazon_Prime_Visualizations", exist_ok=True)

# Create Histogram
plt.figure(figsize=(8,5))
plt.hist(prime_df['imdb_score'], bins=20, edgecolor='black')

plt.title('IMDb Score Distribution')
plt.xlabel('IMDb Score')
plt.ylabel('Frequency')

# Save the image
plt.savefig(
    "Amazon_Prime_Visualizations/IMDb_Score_Distribution.png",
    dpi=300,
    bbox_inches='tight'
)

# Display the graph
plt.show()
```

# Step 4: Top 10 Highest Rated Titles

```python
import os
import matplotlib.pyplot as plt

# Create folder if it doesn't exist
os.makedirs("Amazon_Prime_Visualizations", exist_ok=True)

# Top 10 movies by IMDb score
top_movies = prime_df[['title', 'imdb_score']]

top_movies = top_movies.sort_values(
    by='imdb_score',
    ascending=False
).head(10)

# Plot
plt.figure(figsize=(12,6))
plt.bar(top_movies['title'], top_movies['imdb_score'])

plt.title("Top 10 Movies by IMDb Score")
plt.xlabel("Movie Title")
plt.ylabel("IMDb Score")

# Rotate x-axis labels
plt.xticks(rotation=45, ha='right')

# Save image
plt.savefig(
    "Amazon_Prime_Visualizations/Top_10_Movies_IMDb_Score.png",
    dpi=300,
    bbox_inches='tight'
)

# Show plot
plt.show()
```

**Insight**

Best-rated content on platform.

# Step 7: Top Production Countries

Your country column is stored as a list-like string.

```python
import os
import matplotlib.pyplot as plt

# Create folder if it doesn't exist
os.makedirs("Amazon_Prime_Visualizations", exist_ok=True)

# Extract countries
countries = (
    prime_df['production_countries']
    .str.strip("[]")
    .str.replace("'", "", regex=False)
    .str.split(',')
    .explode()
    .str.strip()
)

# Plot
plt.figure(figsize=(10,6))

countries.value_counts().head(10).plot(
    kind='bar',
    color='skyblue',
    edgecolor='black'
)

plt.title('Top Production Countries')
plt.xlabel('Country')
plt.ylabel('Number of Titles')

# Save image
plt.savefig(
    "Amazon_Prime_Visualizations/Top_Production_Countries.png",
    dpi=300,
    bbox_inches='tight'
)

# Show plot
plt.show()
```

**Insight**

Which countries contribute most content?

# Step 8: Top Genres

```python
import os
import matplotlib.pyplot as plt

# Create folder if it doesn't exist
os.makedirs("Amazon_Prime_Visualizations", exist_ok=True)

# Extract genres
genres = (
    prime_df['genres']
    .str.strip("[]")
    .str.replace("'", "", regex=False)
    .str.split(',')
    .explode()
    .str.strip()
)

# Create figure
plt.figure(figsize=(10,6))

genres.value_counts().head(10).plot(
    kind='bar',
    color='skyblue',
    edgecolor='black'
)

plt.title('Top Genres')
plt.xlabel('Genre')
plt.ylabel('Number of Titles')

# Save image
plt.savefig(
    "Amazon_Prime_Visualizations/Top_Genres.png",
    dpi=300,
    bbox_inches='tight'
)

# Show graph
plt.show()
```
